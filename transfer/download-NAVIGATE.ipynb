```python
import ixmp4
import pyam
from pathlib import Path
```

```python
platform = ixmp4.Platform("navigate")
```

```python
df = pyam.read_ixmp4(
    platform,
    scenario="PEP*",
)   
```

```python
df.filter(model="REMIND-MAgPIE 3.2-4.6").to_excel("raw/NAVIGATE/REMIND-PEP.xlsx")
```

```python
connection = pyam.iiasa.Connection("navigate")
```

```python
scenarios = connection.index().reset_index()
```

```python
list()
```

```python
scenarios[[i.startswith("PEP") for i in scenarios.scenario]]
```

```python
for i, (model, scenario, version) in connection.index().reset_index().iterrows():
    if "SUP" in scenario:
        try:
            connection.query(model=model, scenario=scenario).to_excel(f"raw/NAVIGATE/{model}-{scenario}.xlsx")
            print(f"Downloaded {model} - {scenario}")
        except:
            print(f"Failed {model} - {scenario}")
```

```python
runs = connection.index(default_only=True).reset_index()
```
